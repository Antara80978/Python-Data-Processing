```python
# Creating a list of student records
students = [
    {"student_id": 101, "name": "Aarav", "department": "IT", "marks": 78},
    {"student_id": 102, "name": "Diya", "department": "Computer", "marks": 85},
    {"student_id": 103, "name": "Rohan", "department": "IT", "marks": 92},
    {"student_id": 104, "name": "Meera", "department": "Computer", "marks": 67},
    {"student_id": 105, "name": "Kabir", "department": "IT", "marks": 88},
    {"student_id": 106, "name": "Isha", "department": "EXTC", "marks": 74},
    {"student_id": 107, "name": "Vivaan", "department": "Computer", "marks": 95},
    {"student_id": 108, "name": "Anaya", "department": "IT", "marks": 81},
    {"student_id": 109, "name": "Arjun", "department": "EXTC", "marks": 69},
    {"student_id": 110, "name": "Sara", "department": "Computer", "marks": 90}
]
```

```python
# Display student records
print("Student Records:")
for student in students:
    print(student)
```

```python
# Function to calculate summary statistics
def calculate_summary(records):
    if not records:
        return None

    marks = [student["marks"] for student in records]

    total = sum(marks)
    average = total / len(marks)
    highest = max(marks)
    lowest = min(marks)

    passed = [s for s in records if s["marks"] >= 40]
    topper = max(records, key=lambda s: s["marks"])

    return {
        "total_students": len(records),
        "total_marks": total,
        "average_marks": round(average, 2),
        "highest_marks": highest,
        "lowest_marks": lowest,
        "passed_students": len(passed),
        "failed_students": len(records) - len(passed),
        "pass_percentage": round(
            (len(passed) / len(records)) * 100, 2
        ),
        "topper": topper["name"]
    }
```

```python
# Function to filter students
def filter_students(records, minimum_marks):
    return [
        student for student in records
        if student["marks"] >= minimum_marks
    ]
```

```python
# Function to display records neatly
def display_records(records):
    for student in records:
        print(
            f'ID: {student["student_id"]}, '
            f'Name: {student["name"]}, '
            f'Department: {student["department"]}, '
            f'Marks: {student["marks"]}'
        )
```

```python
# Calculate and display summary
summary = calculate_summary(students)

print("\nSummary Statistics:")
for key, value in summary.items():
    print(f"{key.replace('_', ' ').title()}: {value}")
```

```python
# Filter students scoring 80 or above
print("\nStudents scoring 80 or above:")
display_records(filter_students(students, 80))
```
